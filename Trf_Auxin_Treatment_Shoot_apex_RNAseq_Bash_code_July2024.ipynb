```bash
#As this data is RNAseq I need to align to the NAM5 genome using HiSAT2 which is good at accounting for splice junctions\
#As it is RNAseq data and we want to account for splice junctions we use HiSAT2 to align the trimmed reads to NAM5 genome
find ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads -name "*_R1.fq.gz" | while read f; do
    STEM=$(basename "${f}" _R1.fq.gz)
    hisat2 --min-intronlen 20 --max-intronlen 20000 \
    -p 4 -x ./Documents/Databases/Maize_genome_NAM5/Zm-B73-REFERENCE-NAM-5.0_chr\
    -1 ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads/"${STEM}"_R1.fq.gz \
    -2 ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads/"${STEM}"_R2.fq.gz \
    | samtools view -bS - \
    > ./Desktop/TRF_AUXIN_RNAseq_July2024/aligned_reads/"${STEM}".bam \
    2> ./Desktop/TRF_AUXIN_RNAseq_July2024/HiSAT2_report/"${STEM}"_summary.txt
done
```

```bash
#As this data is RNAseq I need to align to the NAM5 genome using HiSAT2 which is good at accounting for splice junctions\
#As it is RNAseq data and we want to account for splice junctions we use HiSAT2 to align the trimmed reads to NAM5 genome
mkdir ./Desktop/TRF_AUXIN_RNAseq_July2024/aligned_reads
mkdir ./Desktop/TRF_AUXIN_RNAseq_July2024/HiSAT2_report
find ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads -name "*_R1.fq.gz" | while read f; do
    STEM=$(basename "${f}" _R1.fq.gz)
    hisat2 --min-intronlen 20 --max-intronlen 20000 \
    -p 4 -x ./Documents/Databases/Maize_genome_NAM5/Zm-B73-REFERENCE-NAM-5.0_chr\
    -1 ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads/"${STEM}"_R1.fq.gz \
    -2 ./Desktop/TRF_AUXIN_RNAseq_July2024/trimmed_reads/"${STEM}"_R2.fq.gz \
    | samtools view -bS - \
    > ./Desktop/TRF_AUXIN_RNAseq_July2024/aligned_reads/"${STEM}".bam \
    2> ./Desktop/TRF_AUXIN_RNAseq_July2024/HiSAT2_report/"${STEM}"_summary.txt
done
```

```bash
# run featureCounts on the NAM5 aligned files, with PE settings on my iMAC (April 2022) (using paired end settings)
for f in ./Desktop/TRF_AUXIN_RNAseq_July2024/aligned_reads/*.bam; do
  STEM=$(basename "${f}" .bam)
  featureCounts -T 4 \
      -a ./Documents/Databases/Maize_genome_NAM5/Zm-B73-REFERENCE-NAM-5.0_Zm00001eb.1.gtf \
     -p -t exon -g gene_id \
      -o ./Desktop/TRF_AUXIN_RNAseq_July2024/featureCounts/${STEM}.counts.txt \
      ./Desktop/TRF_AUXIN_RNAseq_July2024/aligned_reads/${STEM}.bam
done
```
